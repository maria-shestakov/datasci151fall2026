# **Warning**

### If you try to create a new variable using the assignment operator with the **entire** list, it will not make a distinct copy! They will be two different names for the same object!

### This next cell shouldn't be very surprising...

```python
# we'll make a list where each entry is a floating-point number
list_of_floats = [2.3, -1.0, 5.1, 9.9]

# Here, we are grabbing *one* element of the list, so a new variable is made
new_variable = list_of_floats[2]

print(new_variable)

list_of_floats[2] = 0.0

print(list_of_floats)
print(new_variable)
```

### But the behavior of this next cell is a little surprising at first 

```python
list_of_floats = [2.3, -1.0, 5.1, 9.9]

# Here, we are using the *entire list* with the assignment operator, 
# so we are NOT making a distinct new variable

list_of_numbers = list_of_floats

print(list_of_numbers)

list_of_floats[2] = 0.0

print(list_of_numbers)
```

```python
# To get the expected result, make a copy: 

list_of_floats = [2.3, -1.0, 5.1, 9.9]

list_of_numbers = list_of_floats.copy()

print(list_of_numbers)

list_of_floats[2] = 0.0

print(list_of_numbers)
```

# <span style="color:red; font-size:30px"> NumPy Arrays</span>

<font size = "3">

- We saw that the "+" operator *concatenates* (or combines) two lists. This is also true for other sequence objects, like tuples and strings.

- What if we actually want to add the elements in two sequences entry by entry? We can do this using **NumPy** arrays.

- The NumPy library (NUMerical PYthon) is a library that is widely used for numerical computations (computations with floating-point numbers)

- It comes with a bunch of mathematical functions - **it basically makes the "math" library obsolete**

- It also comes with its own sequence object, the NumPy array

```python
# import the library. "np" is the standard nickname
import numpy as np

# The most common way to create a Numpy array is to convert a list 
# into a numpy array.

list_of_numbers = [5.6, 2.9, 9.7, -10, 0, 3]

array_of_numbers = np.array(list_of_numbers)

print(type(list_of_numbers))
print(type(array_of_numbers))
print(array_of_numbers) #prints numbers
display(array_of_numbers) #makes it clear it is an array 
```

```python
# Usually, we would create the numpy array in a single line, without
# defining a separate list first

array_of_numbers = np.array( [5.6, 2.9, 9.7, -10, 0, 3] )


# you can access elements of arrays in the same way as lists:
print(array_of_numbers[0])
print(array_of_numbers[4])
```

<font size = "3">

- NumPy arrays are primarily useful for storing numbers. You *can* make a NumPy array with strings or other objects, but that is unusual.

- One of the benefits of NumPy is that all operations on arrays are numerical, unlike lists

```python
list_1 = [2, 3, 5, 11]
list_2 = [1, -4, 2, 0]

# The "+" operation concatenates the lists
print(list_1 + list_2)
```

```python
array_1 = np.array([2, 3, 5, 11])
array_2 = np.array([1, -4, 2, 0])

# The "+" operation adds the elements of the two arrays together
print(array_1 + array_2)
```

### The + operation performs element-by-element addition between two arrays of the **same size**. Mathematically, we view NumPy arrays as *vectors*, specifically column vectors.

$\text{An example where both arrays/vectors have 4 elements}$

$ a + b = \begin{pmatrix} a_1 \\ a_2 \\ a_3 \\ a_4 \end{pmatrix} +
          \begin{pmatrix} b_1 \\ b_2 \\ b_3 \\ b_4 \end{pmatrix} =
          \begin{pmatrix} a_1 + b_1 \\ a_2 + b_2 \\ a_3 + b_3 \\ a_4 + b_4 \end{pmatrix}$


### You can also use the "+" operator with a Numpy Array and a single number

$\text{For example:}$

$ a + 2 = \begin{pmatrix} a_1 \\ a_2 \\ a_3\end{pmatrix} + 2 = \begin{pmatrix} a_1 + 2 \\ a_2 + 2 \\ a_3 + 2\end{pmatrix}$

```python
# example

a = np.array([3, 11, -2.2])
print(a + 2)
```

```python
b = [3,11,-2.2]
#print(b + 2) -> doesn't work with regular lists 
print(b + [2]) #adds 2 as an element in the list 
```

# <span style="color:red; font-size:30px"> Plotting functions using NumPy arrays</span>

<font size = "4">

- NumPy arrays are very convenient for plotting functions, because you can use them within mathematical expressions as if they were numbers

- For example, suppose we want to create a scatter plot of the function 

$$ f(x) = x^3 - 2x + \frac{1}{2} $$
&ensp; &ensp; &ensp; at equally spaced points over the interval $-1.5 \leq x \leq 1.5$.

```python
# Create list of equally spaced points from -1.5 to 1.5
# We'll choose points 0.25 apart
x_vals = [-1.5, -1.25, -1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1, 1.25, 1.5]

# Convert to numpy array
x_vals = np.array(x_vals)

# Create new numpy array with the function evaluated at the same points
y_vals = x_vals**3 - 2*x_vals + 0.5

# Use plt.scatter to create scatter plot
plt.scatter(x = x_vals, y = y_vals)
plt.show()
```

<font size = "4">

We can generate a line plot by using `plt.plot` instead of `plt.scatter`:

```python
# Create list of equally spaced points from -1.5 to 1.5
# We'll choose points 0.25 apart
x_vals = [-1.5, -1.25, -1, -0.75, -0.5, -0.25, 0, 0.25, 0.5, 0.75, 1, 1.25, 1.5]

# Convert to numpy array
x_vals = np.array(x_vals)

# Create new numpy array with the function evaluated at the same points
y_vals = x_vals**3 - 2*x_vals + 0.5

# Use plt.scatter to create scatter plot
plt.plot(x_vals, y_vals) #does NOT use x=, y=
plt.xlabel("$x$", fontsize = 16)
plt.ylabel("$y$", fontsize = 16)
plt.title("Graph of $y = x^3 - 2x + \\frac{1}{2}$", fontsize = 12)
plt.show()
```

<font size = "4">

The plot above doesn't look "smooth". We can get a better graph by using more points.

- Instead of typing them out manually, we will use `np.linspace` to generate a Numpy array automatically.

```python
start = -1.5      # points start at -1.5
stop = 1.5        # points stop at 1.5
num_points = 50   # will generate 50 evenly spaced points

x_vals = np.linspace(start, stop, num_points)   # this creates numpy array - don't need to convert it

# Create new numpy array with the function evaluated at the same points
y_vals = x_vals**3 - 2*x_vals + 0.5

# Use plt.scatter to create scatter plot
plt.plot(x_vals, y_vals)
plt.xlabel("$x$", fontsize = 16)
plt.ylabel("$y$", fontsize = 16)
plt.title("Graph of $y = x^3 - 2x + \\frac{1}{2}$", fontsize = 12)
plt.show()
```

<font size = "4">

- You can also use Numpy arrays together with Numpy mathematical functions

- For example, to make a scatter plot of the function

$$ g(t) = \mathrm{e}^{t^2} - \sqrt{t} + 1$$

&ensp; &ensp; &ensp; at equally spaced points over the interval $0 \leq t \leq 0.5$, we can do:
